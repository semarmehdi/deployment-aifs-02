### .post pour la prediction 

```python
import pandas as pd
filename = "https://full-stack-assets.s3.eu-west-3.amazonaws.com/Deployment/ibm_hr_attrition.xlsx"
df_a_pred = pd.read_excel(
    filename,
    index_col=0,
)
echantillon = df_a_pred.sample(10, random_state=0).drop('Attrition', axis=1)
echantillon
```

```python
url_base = "http://localhost:8001"
endpoint = "/predict-attrition"

full_url = url_base + endpoint

payload = echantillon.to_json()


reponse = requests.post(full_url, json=payload)
reponse
```

```python
reponse.content
```

```python
une_ligne = echantillon.head(1)
une_ligne
```

```python
une_ligne.values
```

```python
une_ligne.to_dict()
```

```python
une_ligne_values = une_ligne.values[0]
une_ligne_keys = une_ligne.columns
une_ligne_dict = {}
for value, key in zip(une_ligne_values, une_ligne_keys):
    une_ligne_dict[key] = value


url_base = "http://localhost:8001"
endpoint = "/predict-attrition"

full_url = url_base + endpoint

payload = une_ligne_dict


reponse = requests.post(full_url, json=payload)
reponse
```
